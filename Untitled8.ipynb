```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx"

df = pd.read_excel(url)

print(df.head())
print(df.shape)
print(df.columns)
```

```python
df = pd.read_excel("/content/Online Retail.xlsx")
```

```python
df = df.dropna(subset=["CustomerID"])

df = df[~df["InvoiceNo"].astype(str).str.startswith("C")]

df = df[(df["Quantity"] > 0) &
        (df["UnitPrice"] > 0)]

df = df.drop_duplicates()

print("Cleaned data shape:", df.shape)
print(df.head())
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

rfm = df.groupby("CustomerID").agg(
    Recency=("InvoiceDate",
             lambda x: (reference_date - x.max()).days),

    Frequency=("InvoiceNo", "nunique"),

    Monetary=("TotalPrice", "sum")
)

print(rfm.head())
print("Total customers:", len(rfm))
```

```python
# Reduce skew in the data
rfm_log = np.log1p(rfm)

# Put features on a similar scale
scaler = StandardScaler()
X = scaler.fit_transform(rfm_log)

print("Prepared data shape:", X.shape)
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm["KMeans_Cluster"] = kmeans.fit_predict(X)

print(rfm.head())
print(rfm["KMeans_Cluster"].value_counts().sort_index())
```

```python
inertia = []

for k in range(2, 8):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X)
    inertia.append(model.inertia_)

plt.plot(range(2, 8), inertia, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
dbscan = DBSCAN(eps=1.5, min_samples=10)

rfm["DBSCAN_Cluster"] = dbscan.fit_predict(X)

print(rfm["DBSCAN_Cluster"].value_counts().sort_index())
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

plt.figure(figsize=(8, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=rfm["KMeans_Cluster"],
    cmap="viridis",
    s=15
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Customer Segmentation using PCA")
plt.show()

print("Variance explained:", pca.explained_variance_ratio_.sum())
```

```python
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30,
    init="pca",
    learning_rate="auto"
)

X_tsne = tsne.fit_transform(X)

plt.figure(figsize=(8, 5))

plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=rfm["KMeans_Cluster"],
    cmap="viridis",
    s=15
)

plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.title("Customer Segmentation using t-SNE")
plt.show()
```

```python
summary = rfm.groupby("KMeans_Cluster")[
    ["Recency", "Frequency", "Monetary"]
].mean().round(2)

print(summary)
```

```python
labels = rfm["KMeans_Cluster"]

score = silhouette_score(X, labels)

print("Silhouette Score:", round(score, 3))
```

```python
labels_db = rfm["DBSCAN_Cluster"]
mask = labels_db != -1

if len(set(labels_db[mask])) >= 2:
    print("DBSCAN Silhouette:",
          round(silhouette_score(X[mask], labels_db[mask]), 3))
else:
    print("Not enough clusters for silhouette score.")
```
